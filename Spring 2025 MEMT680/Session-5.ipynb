### Carbonaceous materials: hard carbon to graphite

Lithium metal negative electrodes have the lowest reduction potential compared with any other metal (-3.04 V versus standard hydrogen electrode) as well as a high theoretical specific capacity, and thus enable a high voltage, high energy density battery. However, the problem lies in its reactivity - initial commercialization efforts by Moli Energy in the 1980s led to a product recall due to recurring fires. 

[ref: 10.1149/1.2085950]

An alternative to lithium metal is the insertion of lithium ions into a host structure, such as carbon. However, early studies were not able to insert and de-insert lithium from carbon (either petroleum coke or graphite). Most studies used propylene carbonate (PC) solvent, which causes co-intercalation of the solvent along with continuous solvent decomposition. 



[ref: doi.org/10.1021/cm901452z]

In 1990, Jeff Dahn's research group found that the addition of ethylene carbonate (EC) as a co-solvent significantly improves the reversible charging and discharge of lithium into graphite. It was thought that the co-intercalation of PC led to exfoliation of graphite and continuous solvent decomposition; this process could be stopped by the reduction of EC forming a stable SEI layer. 

The differences in the voltage curves between coke and graphite are shown:



In short, using a combination of EC and PC enabled long-term cyclability of lithium into and out of graphite. 

[ref: doi.org/10.1149/1.2086855]

Graphite intercalation occurs via staging. Initially, lithium intercalates into graphite randomly, in a "liquid-like" state, called stage 1L. A first order phase transition occurs to stage 4, where lithium fully occupies every 4th layer. This is followed by a transition to stage 3, to stage 2L and stage 2. This all happens within the first 50% of charging / SOC. Now lithium occupies every other layer. Lastly, there is a phase transition to stage 1 to form LiC6. At such low potentials of 0.25 to 0.01 V versus Li/Li+, most electrolytes are not thermodynamically stable and hence will decompose to form SEI. 



[ref: https://pubs.rsc.org/en/content/articlehtml/2020/se/d0se00175a]

- There are two types of graphite used in electrodes: synthetic and natural. Natural graphite is mined and refined, synthetic graphite is artificially produced thru high temperature processing. Synthetic graphite is more widely used because it tends to have higher purity and higher density, but is more expensive and has more carbon emissions resulting from its processing. 

### Silicon

Lithium can be alloyed with a number of metals, resulting in a specific capacity higher than that of graphite. Specifically, lithium forms an alloy with silicon resulting in nearly 4 atoms of lithium per atom of silicon ($Li_{15}S_4$), which comes out to a specific capacity of ~3579 mAh/g. Both amorphous and crystalline silicon has been observed. 

- However, this alloying (charging) and dealloying (discharging) process leads to significant volume change, leading to mechanical cracking and disintegration of the electrode. 

- In 1995, Wilson and Dahn prepared carbons with nanodispersed silicon, showing reversible capacity. Their result demonstrated that you could make nanosized silicon particles well dispersed in graphite (though only up to 4%) exhibiting an increase in capacity on the first cycle (though it seems in this case most of the added capacity due to the silicon is lost by the fourth cycle)

[Ref: doi.org/10.1149/1.2043994]

- Other studies in the 2000s also indicated that nano-structured silicon (particles, films, wires, etc.) mitigates the mechanical stresses induced upon lithiation/delithiation. 
[Ref: https://iopscience.iop.org/article/10.1149/1.1596917]

- In 2008, Amprius was founded to try and use a silicon nanowire approach. 



- In 2011, Sila Nano was founded to make nano-scale silicon powders. Their approach instead focused on scaling: ensuring that the way the silicon is processed is optimized for manufacturing scaleup. They have been targeting ~50% silicon anodes and as of late 2023, have started supplying their silicon materials to automotive OEMs. 

### Li metal 

Can we really double the energy density by switching graphite with lithium metal? Remember that it's not just the negative electrode, but everything else that also counts towards cell level energy density. There should be minimal excess electrolyte, and minimal excess lithium metal. This includes:

- enlarging the cathode capacity to match that of the anode
- using minimal excess electrolyte (less than 3 g/Ah)
- using minimal excess lithium metal (ideally "anode-free")



[ref: https://www.nature.com/articles/s41560-019-0338-x]

Aside from the impact that current density has on lithium metal electrodeposition, changing the electrolyte composition has a significant impact on lithium metal battery rechargeability. LiFSI has a higher ionic conductivity than LiPF6, and DME solvent has a lower reduction potential than the carbonate solvents - but these properties by themselves do not explain why electrolytes with ether solvents usually exhibit higher discharge/charge efficiency. 1M LiFSI DME has low CE%, whereas 4M LiFSI DME presents a very high CE%. In 2015, Qian et al. find that such high concentration electrolytes (4M LiFSI DME) results in over 98% stripping/plating efficiency, one of the highest measurements to date. The attribute this to the high salt concentration reducing the free solvent availability in the electrolyte, minimizing the reactivity between free DME solvents and lithium metal. On the downside, ethers are only stable up to around 4 V, and are therefore not compatible with NMC, LCO, and other higher voltage cathodes. Further, such a high concentration reduces the ionic conductivity of the electrolyte. 

Later in 2018, an electrolyte at 1 M salt concentration was found to perform even better than the high concentration electrolyte. Chen et al. use 1M LiFSI DMC with BTFE co-solvent and call it a "localized high concentration electrolyte" (LHCE). They speculate, using evidence from DFT calculations and NMR, that LiFSI preferentially interacts with DMC. In other words, BTFE solvents take up volume in the electrolyte but do not coordinate with the salt, leading to localized regions of highly coordinated LiFSI-DMC. In these regions, then, the electrolytes behaves like the high concentration case where minimal free solvents are present. [ref: doi.org/10.1002/adma.201706102]

Electrolytes impact SEI formation and the stability of lithium metal batteries, but to date, there is no universal understanding of why certain electrolytes work better than others. Empirically, ethers are better than carbonates, and higher salt concentrations (or local concentrations) are also better.

### Hard carbon for Na-ion  

Na-ion batteries are a promising alternative to Li-ion due to the low cost and abundance of sodium. Sodium has a reduction potential almost as negative as lithium. However, sodium ions are too large to intercalate into graphite, and instead is more compatible with hard carbon, which is an amorphous material. Early studies implied that an adsorption-like process was occurring in hard carbons. 

- Hard carbon consists of randomly oriented graphitic domains with higher interlayer spacing, with randomly closed pores in between the graphitic crystallites. 

- A "house of cards" model suggests that sodium ions insert into parallel or nearly parallel layers of the randomly stacked hard carbon material, and then inserts into nanoporous regions between stacked layers (in an adsorption-like process). 

- Why can't Na ions intercalate into graphite? It is just the larger size? But K ions can intercalate, and they are even bigger than Na ions..one explanation is the relatively weak sodium ion - carbon bond

The total amount of lithium ions can be considered the "active lithium inventory." As the battery is cycled at fast charge rates, more and more lithium ions are temporarily or permanently "lost" due to side reactions such as lithium metal deposition, instead of intercalation. This results in "lost lithium inventory" that causes a decline in charge capacity. Some of the lithium metal deposits are not necessarily lost, as they are still electrically connected and can be stripped off upon discharge. Some may be electrically disconnected, or "dead", but subsequent cycling may "revive" it and contribute to slight increases in capacity. 

Aside from optical microscopy which requires a custom designed optical cell, other methods to detect lithium plating have included X-ray diffraction and X-ray tomography (both requiring advanced, high resolution synchrotron sources), nuclear magnetic resonance spectroscopy, ultrasonic transmission.

### What about Li metal anode dynamics?

Using an "anode-free" form factor involves simply taking out the graphite. In theory, this removes a manufacturing step and lowers cost. All the lithium inventory comes from the cathode. 



[Ref: https://www.nature.com/articles/s41560-020-0668-8/figures/1]

Repeated lithium metal electrodeposition and stripping leads to significant changes in morphology. Usually, a higher uniaixial force ("stack pressure") is applied to maintain a uniform morphology. Higher pressures are correlated with higher Coulombic efficiencies, as less lithium is lost. Still, loss of active lithium gradually occurs. Electrolyte is gradually consumed, forming progressively more SEI and porous lithium morphology. 



Typically, lithium metal plating increases the cell volume leading to an increase in cell pressure during charge.
