# Solid-state electrolytes

Solid-state electrolytes are ionically conductive, electronically insulating solids that are, in theory, safer than flammable organic liquid electrolytes. 



## Solid ion conductors

Aside from liquid electrolytes, lithium ions can be mobile within a solid-state material. Within crystalline compounds, lithium ions can migrate from one vacant atomic site to another. 



The mobility of the ion is its ability to move through the lattice. The exact mechanism is complicated, and dependent on the type of solid electrolyte. The basic version is that one ion hops from one site to another. Multiple ions can also hop simultaneously to their nearest sites, in so-called "lithium-ion superionic conductors" or LISICON. 

[ref: https://www.nature.com/articles/ncomms15893]

At larger length scales, grain boundaries are sure to play a role as well. Solid electrolytes with pores lead to more tortuous paths for ion conduction. Interfaces also matter. Poorly contacting interfaces lead to poor ion conduction. 

Therefore, the understanding of ion conduction in solid electrolytes is a *multi-scale* problem. 

- at the atomic scale, ion hopping kinetics from one site to another
- at the microscale, conduction occurs differently through grain boundaries, in the bulk crystallite, or via amorphous phases 
- at the mesoscale, overall particle-particle or electrode/electrolyte contact, as well as current density distribution, impacts spatial heterogeneities in ion transport
- at the cell level, the overall stack pressure (uniaxial forced imposed on the cell), operating temperature, cycling conditions, cell form factor all play a role 



[ref: https://www.nature.com/articles/s41563-019-0431-3/figures/2]


In the 1960s, sodium beta-alumina was a solid material discovered to have ionic conductivity for sodium ions. This led to the development of a high temperature Na-S battery by Ford and the "ZEBRA" battery. In the 1970s, solid polymer electrolytes were developed and in 1983, lithium phosphorus oxynitride (LiPON) was developed as a thin film solid state battery. 

In the 2000s, other classes of solid ion conductors emerged. LGPS (argyrodites) in 2011 exhibited an ionic conductivity equivalent to that of organic liquid electrolytes (12 mS/cm). Also, garnets (LLTO, LLZO), perovskites (LLTO), NASICON. 

All of these solid electrolytes have issues. For instance, while sulfide electrolytes have the highest ionic conductivity, they will react irreversibly with lithium metal. Garnets are prone to brittle fracture and require very high temperatures (over 1300degC) to sinter and densify. In general, it remains difficult to manufacture solid electrolytes to a thickness that would be desirable for a high energy density  battery (less than 20 microns, whereas most academic research investigates pellets that are at least 0.5mm thick). There are also all the challenges that come with cycling them - maintaining particle contact between the electrolyte and the active material despite the significant volume changes.

## Garnets

A common garnet solid electrolyte has the composition: lithium lanthanum zirconium oxide ($Li_7La_3Zr_2O_{12}$). LLZO was first reported by Murugan et al in 2007, and exhibits high ionic conduction (0.3 mS/cm) and high thermal and chemical stability against battery electrodes. As with other solid electrolytes, these are usually made by solid-state reactions. Stoichiometric amounts of lithium hydroxide, lanthanum oxide and zirconium oxide are obtained and ball milled to reduce particle size. The particles are then pressed into dense pellets and annealed. 

[ref: https://doi.org/10.1002/anie.200701144]



- External force ('stack pressure') has a significant impact on internal resistance
- Non-uniform lithium stripping at high current densities will lead to loss of contact, leading to current constriction
- This can be partially alleviated by applying a higher stack pressure so that the relatively soft lithium creeps 

For more info, see literature by Jurgen Janek or Jeff Sakamoto 

[ref: doi.org/10.1021/acsami.9b02537]

## Sulfides

Li-ion conductive sulfide glasses were synthesized back in the 1980s. Various sulfide solid electrolytes have measured ionic conductivities up to 10 mS/cm (around the same as conventional liquid electrolytes). Another advantage over garnets is the capable of sulfides to be sintered at room temperature via cold pressing, and their lower elastic moduli. Examples include: $Li_{10}GeP_2S_{12}$ (LGPS, or LISICON - lithium superionic conductor), $Li_7P_3S_{11}$ (LPS), and $Li_6PS_5Cl$ (LPSCl, or argyrodite). 



[ref: https://www.frontiersin.org/articles/10.3389/fenrg.2016.00025/full]

However, sulfides are reactive to moisture and can produce hydrogen sulfide ($H_2S$) via hydrolysis. 

Sulfides will also react with lithium metal, forming $Li_2S$, $LiCl$, and $Li_xP$. 

## Halides 



First synthesized in 1990s, reevaluated by Asano et al. in 2018. General composition is $Li_3MX_6$ where X = Cl, Br or I. For instance, lithium indium halide and yttrium halide have been investigated as well as various derivatives (LiInZrCl). They appear to be oxidatively stable but reductively unstable against lithium metal. 

As both sulfides and halides react with lithium metal, they are generally tested against a lithium-indium alloy. Since halides are not reductively stable, many are tested by using an additional separator that is sandwiched between the lithium and the halide, such as a thiophosphate solid electrolyte. As long as this other separator forms a kinetically stable interphase, this avoids a continuous reaction. 

## Measuring ionic conductivity

A  two electrod cell with symmetric ion blocking electrodes (e.g. stainless steel, resulting in no charge transfer at the electrode/electrolyte interfaces) can be used to measure the ionic conductivity of the electrolyte: $\sigma _{ion}\ =\ l/R_{total}A$

## Void formation

On discharge when metal is stripped from the negative electrode, voids can form at solid/solid interfaces because of poor wettability and loss of contact. This is due to the relatively slow creep of lithium towards the interface relative to lithium depletion and migration into the electrolyte. The voids are correlated with an increase in cell resistance and measured cell voltage. 

- Since $R_{int}\ =\ \rho l / A$ as the interfacial contact area decreases, the interfacial resistance necessarily increases. This causes an increase in the measured electrode overpotential. The overpotential can be decreased if contact is maintained, which can be done by applying a higher uniaxial force (stack pressure).



[ref: https://www.sciencedirect.com/science/article/pii/S2542435119303083]

- For instance these voids can be observed by synchrotron X-ray tomography in the lithium metal / LSPS interface. For these sulfides, an addition reaction between lithium metal and the electrolyte also leads to the formation of a separate phase that is distinguished from the bulk electrolyte. 

- Along with cracks, evidence of intergranular lithium penetration (at the grain boundaries) is observed from SEM of a fractured electrolyte. 





[ref: https://www.nature.com/articles/s41563-020-00903-2#Sec4]

## Critical current density

The critical current density is defined as the maximum sustainable current density beyond which initiation and propagation of metal dendrites will occur. This was first introduced as a parameter by Virkar et al. for sodium beta alumina interfaces. Usually, the CCD is measured as the current density at which the cell voltage tends to drop to 0 V (by induction of an internal short). 

- The CCD usually increases with increasing temperature and is inversely related to the interfacial resistance. 

- The maximum stripping current density has also been argued to impact short circuiting even if the plating current density is low. Kasemchainan et al. argue this is because fast stripping currents cause current constriction, leading to high local current densities that cause local dendrite initiation and propagation into the electrolyte. 

- For example, the crack visualized below in LLZO is believed to have been caused by the propagation of lithium metal into the solid electrolyte, eventually causing a short circuit at a sufficiently high ion flux (current density). 



[ref: https://www.sciencedirect.com/science/article/pii/S0013468616325609#fig0015]

## Mechanism of lithium penetration through solid electrolytes

But why does lithium penetrate into solid electrolytes? Lithium metal is one of the softest and most ductile metals, with a yield strength of around 1 MPa. 

- The Monroe Newman hypothesis states that metal dendrite growth during electroplating can be suppressed if the electrolyte's shear modulus is high enough. In other words, a sufficiently stiff electrolyte should be able to cause the lithium deposit to deform, and stop dendrite propagation. 

- Despite this, many groups began to observe short circuiting of lithium metal in solid electrolytes. 

- Porz et al. suggest that electrodeposition of metal within a surface crack causes local mechanical stress to build up, and cause the crack to extend. Their experimental evidence of lithium filling into surface cracks and subsequent crack elongation shows that electrolytes with much higher shear modulus can still experience lithium penetration via crack propagation, and that these lithium filaments are not necessarily "dendritic" in nature. [ref: https://onlinelibrary.wiley.com/doi/epdf/10.1002/aenm.201701003?saml_referrer]

- Another conjecture is that some electronic conduction in the electrolyte can cause lithium reduction to occur within rather than right at the interface - causing lithium metal to deposit inside the solid electrolyte [ref: https://www.nature.com/articles/s41563-019-0438-9#Abs1]



[Ref: https://www.sciencedirect.com/science/article/pii/S0378775315304407]

"The membrane is sandwiched between two conductive metal plates, while held in an oven at a modest temperature (800 C) where Lithium evaporation is not significant. An electric field is then applied to the sample to induce FAST sintering. The field can be a D.C. field or an A.C. field. In some examples, the A.C. field is advantageous as a sufficiently high frequency can be selected so that the ionic species do not segregate significantly under the stimulus. Power delivery to the sample must be controlled to avoid excessive joule heating of the material. In some examples, this can be accomplished by initially operating in a constant voltage amplitude mode and switching to constant current once the sintering begins and the impedance of the sample falls. The electrolyte membrane can be sintered to full density in a much shorter time and at a lower temperature than the conventional process. While certain solid state ionic conductors can be sintered in a conventional process by pressing small pellets, which are approximately 10 mm in diameter and 2 mm thick in thickness, known methods for making thin films of garnet based materials are insufficient for battery applications, which require film lateral dimensions of approximately 10 cm and between 100 nm to 50 μm in thickness.

Sintering thin films, particularly films that include garnet (e.g., lithium-stuffed garnet), using applied electrical currents and voltages is inherently challenging. In part, this is related to the resistive heating that occurs in the garnet material when current flows there-through and thereby causes a sintering effect. For example, when electricity is used to sinter garnet, as is done with FAST sintering, the electricity resistively heats and sinters the garnet material primarily where the impedance is the greatest."

### What are the design engineering gaps of solid state batteries?

Quoting from [ref: https://pubs.acs.org/doi/full/10.1021/acsenergylett.1c00445], the following cell-level challenges remain:

Can volume changes be cleverly directed to provide internal compression at the material interfaces to replace large external stack pressures?
- much of the academic literature uses very high external stack pressures (10 MPa or higher) that are not practically attainable in a battery pack. how do we stabilize interfaces without requiring such high stack pressures

Are there other ways to temper the cathode and its internal interfaces to strengthen and resist fracture?
How should composites be designed to maintain their structure and internal connectivity during cycling?
- need to consider not just the interface between the bulk electrolyte and electrode, but the interfaces between each composite cathode particle and the surrounding ion conducting electrolyte material

How do volume changes during cycling affect solid-state battery performance and packaging requirements?
- e.g. changes in thickness from Li plating will cause a change in internal pressure

How can chemical reactions between the cathode and solid electrolyte during densification and cycling be mitigated?
