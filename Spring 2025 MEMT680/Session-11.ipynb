Cylindrical cells: these were the first to be commercialized, and took advantage of existing roll to roll manufacturing equipment for camcorders, tapes, etc. by Sony. The standard size is "18650", referring to the dimensions: 18 mm in diameter, 65.0 mm in height. Recently, larger cells have started to become manufactured: 21700s, 4680s to name a few. Electrode and separator sheets are rolled together and packed into the cylindrical steel can. On the positive end is a venting mechanism containing a fuse that cuts the electrical connection if over-pressurized. When this happens, gas is also allowed to vent out. This is a built in safety mechanism that is meant to mitigate the risk of thermal runaway. 



[ref: https://batteryuniversity.com/article/bu-304-why-are-protection-circuits-needed]

Pouch cells have a polymer casing (usually, it's a thin piece of aluminum with polymer on either side) and are flexible. 

Prismatic cells are similar to pouch cells, but with a hard casing. They are usually built larger. 

### Designing a cell: power vs energy

Power and energy are coupled in Li-ion batteries. To increase energy density (amount of charge stored), you can make electrodes denser and thicker. To increase power density, you need to move ions in and out faster. However, denser and thicker electrodes usually cause ions to move in and out slower. Hence the tradeoff.

### Tesla Model S 

Contains 7,104 Li-ion batteries in 16 modules in series. Each module has 74 cells in parallel with 6 cells in series. So, each module is about 22 V and about 220 Ah. The entire battery pack is then a 350 V pack * 220 Ah = 77 kWh

Now consider the Model Y which contains the new 4680 cells. 4680 cells have a capacity of 26 Ah (!) instead of 3 Ah. If the car continues to require a 350 V pack, about how many cells are needed?
