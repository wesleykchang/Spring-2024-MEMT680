## Battery characterization: non-destructive methods

### Basic measurements: thickness, stress and strain, temperature 

Uniaxial (compressive) pressure is applied on commercial batteries to maintain contact between battery components. For instance, prismatic or cylindrical cells have rigid housings and the electrode stack is compressed inside. This applied stress is a dynamic quantity that changes over battery lifetime because 1) electrodes undergo volumetric expansion and contraction during charge and discharge, and 2) electrodes swell over many cycles and long periods of time as they age. 

- The simplest way to measure these changes in stress are with load cells or thickness gauges, which typically show an increase in stress occurs during charging of a Li-ion battery, followed by a decrease during discharge. 



- Increase in stress over time is caused by SEI growth, structural changes of electrode particles, gas formation, and could cause: separator pore closure due to elastic nature of soft polymer, leading to uneven lithium transport and distribution

- Historically battery researchers focused on the surface chemistry of graphite electrodes, neglecting the equally important effects of mechanical property changes. Over 10 years ago, Arnold et al. (Princeton) began to explore the coupling between chemistry and mechanics:

### X-ray CT 

X-ray CT is now a common technique to image the internal structure of a Li-ion battery. Steel, copper are attenuative to X-rays, aluminum, graphite and separator exhibit low attenuation. 

X-ray nano-CT can reconstruct the microstructures of individual electrode particles as well as failure of the overall cell. This includes visualization of electrode swelling, gas formation, spatial heterogeneities, including failure analysis of cells that experienced thermal runaway. 



[Ref: https://www.sciencedirect.com/science/article/pii/S1369702119303396#s0035]

### Ultrasound 

Ultrasound is another common technique used in geophysics for flaw detection and in biomedical engineering / medical applications. It is a non-destructive method to probe mechanical properties by transmitting sound waves at ultrasonic frequencies (MHz range). 

Sound speed is related to material density and modulus: $c\ =\ \sqrt{E/\rho}$ where E is the elastic modulus and $\rho$ the density. Ultrasonic waves travel through metals at speeds of thousand of meters per second, so it is a rapid technique that can be applied to batteries during cycling (in operando). 

[Ref: https://pubs.rsc.org/en/content/articlelanding/2015/ee/c5ee00111k]




The following video is a simulation (PyClaw acoustics solver) of what happens when an ultrasonic wave passes through a layered battery material.

The most straightforward application of ultrasound is as a non-destructive probe of state-of-charge and state-of-health. Since modulus changes with SOC and SOH, sound speed can track the state of the battery using transducers external to the cell. 



[ref: http://dx.doi.org/10.1149/2.1411712jes]

<video src="JMCA_Movie1.mp4" controls title="Title"></video>

This can be shown by placing ultrasonic transducers inside an X-ray machine to transmit both X-rays and sound waves thru the battery during cycling. This let's us see the cross-sectional change of the battery and the resulting sound speed changes. 

[ref: doi.org/10.1039/d0ta05552b]



Li-ion batteries usually expand on charge and contract on discharge due to graphite lithiation and delithiation. The transmitted sound waves can be carefully analyzed to determine the exact sound transmission speed, and then calculate the modulus. 



The relationship between sound speed and modulus is derived from fundamental acoustic wave equations, where the sound speed is the square root of the modulus divided by the density: $c^2\ =\ M_{eff}/\rho$

Here, $M_{eff}$ is the "effective" modulus, or the average of the moduli of each layer in the battery. Since the sound wave propagates thru all layers, it is difficult to discern between them, so we take an "averaged" property. 
 


By tracking how sound speed, and effective modulus chanage over cycling, we see how the battery evolves dynamically.





### Detection of gas formation and lithium plating with ultrasound

Ultrasonic waves attenuate most at solid/liquid interfaces due to poor sound transmission through air. Therefore, any gas bubbles or voids at an interface are expected to cause near complete loss of an ultrasonic wave signal. This means ultrasound is particularly sensitive to gas formation detection in batteries. 

In the following experiment to demonstrate this, five Li-ion pouch cell batteries were cycled fast at low temperatures, because we know lithium will plate. The point was to see how higher temperatures could react with the plated lithium to form gas. This is meant to simulate a real world event, where one uses an EV in cold climates, and then lets it sit at warmer temperatures (i.e. in a heated garage). 

In short, cycling a battery at 10 degC and then bringing it up to any temperature above room temperature could be undesirable. It was found that the higher the temperature we go, the faster sound signal is lost (and the faster gas formation occurs). 



[ref: https://iopscience.iop.org/article/10.1149/1945-7111/ab6c56/meta]



Therefore, gas formation is ultimately detected via near complete loss of signal due to the poor sound transmission through air. 

This is differentiated from other forms of cell degradation, such as lithium plating and progressive SEI formation. These events will cause the stiffness of the battery to decrease, resulting in a slower sound transmission speed.
