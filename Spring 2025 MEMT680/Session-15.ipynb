### LCOE

The sun doesn't always shine, the wind doesn't always blow - so we need methods to storage excess energy and use it during peak demand times. 

The cost of solar and wind in comparison with conventional energy generation technologies is most widely determined by the "levelized cost of energy" (LCOE) by Lazard. The LCOE is a calculation that takes the total cost of an electricity generation system, such as a module of solar panels, including their initial deployment and installation cost, operation and maintenance, divided by the total amount of electricity generated. This gets a "cost per energy" ($/MWh) that can be compared across different electricity generation technologies. 

[ref: https://www.lazard.com/research-insights/2023-levelized-cost-of-energyplus/]

### LCOS

Deployment of a grid storage technology includes both upfront capital cost and operating costs. Both are important, because grid storage is only feasible if it results in consumer electricity costs comparable to the incumbent. 

Typically, the cost of energy storage technology deployment is quantified via the LCOS (levelized cost of storage), which accounts for the entire lifetime cost and is used for cost comparison between different technologies. For instance, a flow battery that lasts 30 years with a higher upfront cost may have the same LCOS as a different flow battery that lasts only 15 years but with a lower upfront cost.

Examples of grid storage include: pumped hydropower, compressed air, Li-ion batteries, flow batteries (vanadium redox, zinc bromine, iron), lead acid batteries. Each technology has different strengths and weaknesses: initial costs, system lifetime, discharge duration at rated power, response time, etc.



[ref: https://www.sciencedirect.com/science/article/pii/S254243511830583X#sec1]

- Li-ion batteries are cost appropriate for shorter duration discharge times at rated power, because they scale linearly with system size. A 10 MWh Li-ion battery system is comprised of twice as many cells as a 5 MWh system, hence doubling in initial capital cost. With further decrease in Li-ion cost, they are expected to be used even more widely for grid storage. 

- Discharge duration at rated power: the storage duration of a battery system at a specific rated power. For instance, a 10 MWh Li-ion battery system at a rated power of 5 MW has a discharge duration of two hours. 

- Specifically, LFP-based Li-ion batteries and second-life batteries are seen as potentially competitive for grid storage due to lower costs. 

- V2G: vehicle-to-grid. An EV is sitting there for most of its lifetime, but instead could be used as a local form of energy storage to balance the grid. By storing excess energy from the grid when plugged in, and then using that later on during peak demand, an EV consumer could make money from their EV. 

- However, a flow battery at 10 MWh vs 5 MWh just requires a larger tank, or a series of larger tanks, meaning that cost scales asymptotically. Thus, at larger sizes / longer durations at rated power, flow batteries may be more economical. 

- There are also ways to store energy seasonally (many months of duration at rated power), which may be useful for balancing seasonal load differences. 

There are several different use cases for these storage applications:

a) correct sudden frequency and voltage changes in the grid ("ancillary services")
b) correct imbalances between load and generation ("load-leveling")
c) ensure backup power is available during peak demand (i.e. replacement for peaker plants)
d) restore power plant operation after blackout ("black start")
e) seasonal storage 
f) optimize pricing 

### Efficiencies

1) Coulombic efficiency: discharge capacity over charge capacity. This is important for rechargeable batteries for electric vehicles, because we are limited by the total cell capacity and want to ensure we can retrieve all the ions we charge to maximize range on a single charge. However, CE is less important for grid storage, e.g. flow batteries. A lower CE can be compensated for by having excess electrolyte. 

2) Voltaic efficiency: average discharge voltage divided by the average charge voltage. 

3) Round trip energy efficiency: total energy discharged divided by the total energy charged over a full charge and discharge cycle. Typically includes losses from self-discharge (can be non-negligible in some flow battery systems), electrical and heat losses

### Zinc-bromine flow batteries

On charge, zinc is plated as metal on the negative electrode and bromine gas is evolved on the positive electrode. On discharge, zinc metal is stripped and bromine converts into bromide. 

- The bromine is typically sequested with a bromine sequestering agent. 
- Energy is stored via electrolyte in the tank. Power is determined by the size of the cell stack. Unlike Li-ion batteries, flow batteries decouple energy and power. This means if you want a battery storage system that can last longer on a single charge (energy) then you make one with a larger tank. If you want one that can balance the grid quickly by drawing high power, you use a larger cell stack.
